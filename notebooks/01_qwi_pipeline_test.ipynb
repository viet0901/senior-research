# Check the QWI Research Data

This notebook checks the data created by `code/qwi_pipeline.py`. It does not download data. Run `python code/qwi_pipeline.py` from the project root before using this notebook.

```python
from pathlib import Path
import json
import pandas as pd

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "code" / "qwi_pipeline.py").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_FILE = PROJECT_ROOT / "data/interim/qwi/qwi_research.csv"
VALIDATION_FILE = PROJECT_ROOT / "data/interim/qwi/qwi_research_validation.json"

if not DATA_FILE.exists() or not VALIDATION_FILE.exists():
    raise FileNotFoundError(
        "Run 'python code/qwi_pipeline.py' from the project root first."
    )

qwi = pd.read_csv(
    DATA_FILE,
    dtype={"state": "string", "industry": "string", "agegrp": "string"},
)
with VALIDATION_FILE.open(encoding="utf-8") as file:
    validation = json.load(file)

print(f"Loaded {len(qwi):,} rows from {DATA_FILE.relative_to(PROJECT_ROOT)}")
qwi.head()
```

## Basic research checks

The following cell checks the time coverage, state coverage, age groups, duplicate keys, and any failed API requests. Review these results before analysis.

```python
print("Shape:", qwi.shape)
print("Period:", qwi["period"].min(), "to", qwi["period"].max())
print("States/D.C.:", qwi["state"].nunique())
print("Age groups:", sorted(qwi["agegrp"].dropna().unique()))
print("Industries:", qwi["industry"].nunique())
print("Duplicate key rows:", validation["duplicate_key_rows"])
print("Failed requests:", validation["failed_request_count"])
print("Latest common quarter:", validation["latest_common_period"])
```
